```python
# import the necessary libraries

import mysql.connector as sql
import pandas as pd
```

```python
# hostname, database name, username, password
```

```python
db_connection = sql.connect(host = 'localhost',
                           database = 'mydb',
                           user = 'root',
                           password = 'root')
```

```python
db_connection
```

```python
# create a cursor object
```

```python
db_cursor = db_connection.cursor()
```

```python
# execute a query
```

```python
db_cursor.execute('select * from student')
```

```python
table = db_cursor.fetchall()
```

```python
table
```

```python
db_cursor.column_names
```

```python
# create a dataframe
```

```python
df = pd.DataFrame(table, columns= db_cursor.column_names)
```
