```python
class Employee():
    def __init__ (self, first, last, salary):
        self.first = first
        self.last = last
        self.salary = salary
        
    def msg(self):
        print(self.first + ' ' + 'OK')
```

```python
emp1 = Employee('Raj','Gupta',45000)
```

```python
print(emp1.last)
```

```python
emp2 = Employee('Riya','Roy',43000)
```

```python
print(emp2.first)
```

```python
emp1.msg()
```

```python
emp2.msg()
```

```python

Encapsulation=>
u can make a variable private by using __ before it
so others cannot access that particular information


class Employee():
    def __init__ (self, first, last, salary):
        self.first = first
        self.__last = last
        self.salary = salary
        
    def msg(self):
        print(self.first + ' ' + 'OK')


so here ur 'last' name is private variable
wont be able to run=> print(emp1.last)
```

```python
# Inheritance
```

```python
class Employee2(Employee):
    pass

# Employee2 -> Subclass
# Employee -> Superclass
# subclass will inherite from superclass
```

```python
emp3 = Employee2('Kartik','Patil',30000)
```

```python
print(emp3.last)
```

```python
# Create a class
class Person:                        # class name
    def __init__(self, n, a=25):     # constructor
        self.name = n                # instance variables
        self.age = a
    def show(self):                  # method
        print('Name:', self.name)
        print('Age :', self.age)
        
# create the object
p1 = Person('Ajay')  

# call the method
p1.show()
```
