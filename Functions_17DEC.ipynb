```python
# multi parameterized function

def larger(x,y):
    if x > y:
        print('Larger number is ', x)
    else:
        print('Larger number is ', y)
```

```python
larger(10,20)
```

```python
larger(45.67,45.66)
```

```python
# function to print a name multiple number of times

def display(name,count):
    for n in range(count):
        print(name)
```

```python
display('Aditya', 7)
```

```python
# factorial

def factorial(n):
    fact = 1
    for x in range (1, n+1):
        fact = fact * x
    print(fact)
```

```python
factorial(5)
```

```python
# Function returning value

def square(num):
    sq = num * num
    return sq
```

```python
# use return in finding larger value function

def larger(x,y):
    if x > y:
        return x
    else:
        return y
```

```python
larger(45,50)
```

```python
val = larger(30,40)
print('larger is ', val)
```

```python
# multi return

def calc(n):
    sq = n * n
    cu = n ** 3
    return sq, cu
```

```python
calc(10)
```

```python
print('Values are ', calc(10))
```

```python
# Default parameters

def add(x=5, y=10, z=15):
    result = x + y + z
    return result
```

```python
add(x=90, y=20, z=55)
```

#### Lambda Function

```python
square = lambda x: x ** 2

# lambda arguments: expression
```

```python
square(2)
```

```python
# multiplication of 2 numbers

mult = lambda x, y: x * y
```

```python
mult(10,20)
```

```python
# filter

num = [36,56,45,67,21,27]
```

```python
new = filter(lambda x: x < 45, num)
print(list(new))
```

```python
# map  -> gives modified values

new = map(lambda x: x/2, num)
print(list(new))
```

```python
num = [1,2,3,4,5]
```

```python
# factorial
new = map(lambda x: factorial(x) , num)
print(list(new))
```

```python
num = [36,56,45,67,21,27]
new = map(lambda x: x < 45, num)
print(list(new))
```
