```python
import pandas as pd
```

#### import the dataset

```python
df = pd.read_csv('Salary_Data.csv')
x = df['YearsExperience'].values
x = x.reshape(30,1)
y = df['Salary']
```

```python
x.shape
```

#### Holdout method

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x, y, random_state=0,
                                                   test_size= 0.25)
```

#### Build linear regression model

```python
from sklearn.linear_model import LinearRegression
reg = LinearRegression()
reg.fit(x_train,y_train)
print('Training acc: ', reg.score(x_train, y_train))
print('Testing acc: ', reg.score(x_test, y_test))
```

#### K-Fold Method

```python
from sklearn.model_selection import KFold
```

```python
kf = KFold(n_splits= 5, random_state= 0, shuffle= True)
```

```python
for train, test in kf.split(x, y):
    print(train,'\n', test)
```

#### Extract  train and test data

```python
for train, test in kf.split(x, y):
    print('Train:', x[train].flatten())
    print('Test:', x[test].flatten())
```

#### Apply KFold

```python
train_acc = []
test_acc = []

for train, test in kf.split(x, y):
    reg.fit(x[train], y[train])
    print('Train Acc:', reg.score(x[train], y[train]))
    print('Test Acc:', reg.score(x[test], y[test]), '\n')
    train_acc.append(reg.score(x[train], y[train]))
    test_acc.append(reg.score(x[test], y[test]))
```

```python
train_acc
```

```python
len(train_acc)
```

```python
sum(train_acc)/len(train_acc)
```

```python
sum(test_acc)/len(test_acc)
```

#### Apply the kfold to classification

```python
df = pd.read_csv('banknotes.csv')

x = df.drop('Class', axis = 1)
y = df['Class']
```

```python
df.head()
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
dt = DecisionTreeClassifier(random_state= 0)
```

```python
for train, test in kf.split(x, y):
    dt.fit(x.iloc[train, :], y[train])
    y_pred = dt.predict(x.iloc[test, :])
    print('Acc:', accuracy_score(y[test], y_pred))
```

# Leave one out Cross Validation

```python
import warnings
warnings.filterwarnings('ignore')
```

```python
df = pd.read_csv('Salary_Data.csv')
x = df.iloc[:,0:1]
y = df.iloc[:,1]


reg = LinearRegression()
reg.fit(x, y)
print('Acc:', reg.score(x, y))
```

```python
from sklearn.model_selection import LeaveOneOut
```

```python
loo = LeaveOneOut()
```

```python
acc = []
for train, test in loo.split(x, y):
    reg.fit(x.iloc[train, :], y[train])
    print('Training', reg.score(x.iloc[train, :], y[train]))
    acc.append(reg.score(x.iloc[train, :], y[train]))
```
