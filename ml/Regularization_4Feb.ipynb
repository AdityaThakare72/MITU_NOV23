#### Import the libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

#### Read the dataset

```python
df = pd.read_csv('BostonHousing.csv')
```

```python
df.head()
```

```python
df.shape
```

```python
df.isnull().sum()
```

#### Separate input and output

```python
x = df.drop('medv', axis = 1)
y = df['medv']
```

#### cross validation

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, 
                                                   random_state=0,
                                                   test_size=0.25)
```

```python
x_train.shape
```

#### Linear Regression

```python
from sklearn.linear_model import LinearRegression
reg =LinearRegression()
reg.fit(x_train, y_train)
```

```python
print('Training acc: ', reg.score(x_train, y_train))
print('Testing acc: ', reg.score(x_test, y_test))
```

#### Polynomial Regression

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
x_scaled = scaler.fit_transform(x)
```

```python
from sklearn.preprocessing import PolynomialFeatures
poly = PolynomialFeatures(degree = 2)
x_poly = poly.fit_transform(x_scaled)
```

```python
x_train, x_test, y_train, y_test = train_test_split(x_poly, y, 
                                                   random_state=0,
                                                   test_size=0.25)
```

```python
x_train.shape
```

```python
y_train.shape
```

```python
regressor = LinearRegression()
regressor.fit(x_train, y_train)
```

```python
print('Training acc: ', regressor.score(x_train, y_train))
print('Testing acc: ', regressor.score(x_test, y_test))
```

#### Lasso Regression

```python
from sklearn.linear_model import Lasso
regressor = Lasso(alpha = 0.2)

regressor.fit(x_train, y_train)
```

```python
print('Training acc: ', regressor.score(x_train, y_train))
print('Testing acc: ', regressor.score(x_test, y_test))
```

#### Elastic Net

```python
from sklearn.linear_model import ElasticNet
regressor = ElasticNet(alpha=0.1)

regressor.fit(x_train, y_train)

print('Training acc: ', regressor.score(x_train, y_train))
print('Testing acc: ', regressor.score(x_test, y_test))
```
