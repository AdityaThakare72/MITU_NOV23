```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.arange(-5, 5, 0.25)
```

```python
x
```

```python
y = x * 4.67 + 21.25
```

```python
plt.plot(x,y);
```

```python
# Sigmoid

def sigmoid(z):
    val = 1/ (1+ np.exp(-z))
    return val
```

```python
y = sigmoid(x)
```

```python
plt.plot(x,y);
```

```python
# relu -> rectified linear unit
# if value < 0 => 0
# otherwise returns the value

def relu(z):
    if z < 0:
        return 0
    else:
        return z
```

```python
relx = [relu(a) for a in x]
```

```python
plt.plot(x,relx);
```

```python
# softmax

def softmax(z):
    val = np.exp(z)/ np.sum(np.exp(z))
    return val
```

```python
p = [1.45, 2.67, 0.48, 0.95]
```

```python
softmax(p)
```
