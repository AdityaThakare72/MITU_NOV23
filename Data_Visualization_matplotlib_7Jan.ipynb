```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

# Line Plot

```python
x = np.array([12,23,35,46,57])
y = np.array([25,37,48,52,69])
z = np.array([78,64,59,43,26])
```

```python
plt.plot(x,y)
```

```python
plt.plot(x,z)
```

```python
plt.xlabel('First Value')
plt.ylabel('Second Value')
plt.title('Line Plot')
plt.plot(x,y,label='2022', color = 'red', marker = 'o',mfc = 'blue', ls = '--', ms = 7)
plt.plot(x,z,label='2023', color = 'blue', marker = 'o',mfc = 'black', ls = '--', ms = 7)
plt.grid()
plt.legend()
# markers -> -, o, >, <, ^, D, #
```

# Bar Graph

```python
courses = ['java','python','c','c++', 'R']
students = [56,78,68,80,67]
```

```python
plt.xlabel('Course Name')
plt.ylabel('No. of Students')
plt.title('Student Enrollment', fontsize = 20)
plt.bar(courses,students,color = 'black', width = 0.3,
       edgecolor = 'red', linewidth = 3, ls = ':', alpha = 0.4)
plt.grid()
# alpha -> to blur or to faint 
```

# Subplot

```python
plt.figure(figsize=(15,10))

plt.subplot(2,2,2)
plt.xlabel('First Value')
plt.ylabel('Second Value')
plt.title('Line Plot 1')
plt.plot(x,y)

plt.subplot(2,2,1)
plt.xlabel('First Value')
plt.ylabel('Second Value')
plt.title('Line Plot 2')
plt.plot(x,z)
```

# Scatterplot

```python
plt.figure(figsize=(16,9))

plt.xlabel('First Value')
plt.ylabel('Second Value')
plt.title('Scatter Plot')
plt.scatter(x,y, marker = '>')
plt.grid()
```

# Pie Chart

```python
plt.figure(figsize=(10,10))

plt.title('Pie Chart')
mycolors = ['cyan','green','blue','skyblue','red']
plt.pie(students,colors = mycolors, labels = courses, 
        autopct = '%.2f%%',shadow= True, startangle=90, labeldistance=1.1,
       radius = 0.8);
plt.legend()
```
