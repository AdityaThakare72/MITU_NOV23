```python
# read the purchase and selling price from user
# find wheather you got profit or loss AND how much?

purchase_cost = float(input('Enter the purchase cost '))
selling_cost = float(input('Enter the selling cost'))

if purchase_cost < selling_cost:
    print('Profit: ', selling_cost - purchase_cost, 'Rs')
    
else:
    print('Loss: ', purchase_cost - selling_cost, 'Rs')
```

```python
# read three numbers, Find the largest
# cant use 'and' 'or'.... Use nested if only

num1 = float(input('Enter the 1st number '))
num2 = float(input('Enter the 2nd number '))
num3 = float(input('Enter the 3rd number '))

if num1 > num2:
    if num1 > num3:
        print('the largest number is ', num1)
    else:
        print('the largest number is ', num3)
        
else:
    if num2 > num3:
        print('the largest number is ', num2)
        
    else:
        print('the largest number is ', num3)
```

```python
# 2020 -> leap
# 2021 -> not leap
# 2000 -> leap
# 1900 -> not leap -> should be divisible by 4, 100, 400

# check if the year is leap or not

year = int(input('Enter the year'))

if year % 4 == 0:
    if year % 100 == 0:
        if year % 400 == 0:
            print('Century LEAP')
        else:
            print('Not Leap')
    else:
        print('Leap year')
        
else:
    print('Not Leap')
```

```python
'''
read a number from user and do following:
1. find the square
2. find cube
3. check positive or negative
'''

num = int(input('Enter the number '))

print('MENU:\n1.Square\n2.Cube\n3.+ve or -ve')

choice = int(input('Enter the choice: '))

if choice == 1:
    print('Square: ', num * num)
    
elif choice == 2:
    print('Cube: ', num ** 3)
    
elif choice == 3:
    if num > 0:
        print('Number is positive')
    else:
        print('Number is negative')
        
else:
    print('INVALID CHOICE')
```

```python
'''
read two numbers from user and do following:
1. addition
2. floor division
3. num1 expon num2  (num1 ** num2)
'''

num1 = int(input('Enter the 1st number '))
num2 = int(input('Enter the 2nd number '))

print('MENU:\n1.Addition\n2.floor division\n3.num1**num2')

choice = int(input('Enter the choice: '))

if choice == 1:
    print('addition is ', num1 + num2)
    
elif choice == 2:
    print('floor division is ', num1 // num2)
    
elif choice == 3:
    print('expo is ', num1 ** num2)
    
else:
    print('Invalid Choice')
```

```python
''' write a python program to input marks of three subjects 
physics, chemistry, biology. calculate percentage and grade according 
to following. each subject having max 100 marks
 percentage >= 90%: Grade A
 percentage >= 80%: Grade B
 percentage >= 70%: Grade C
 '''

# (marks obtained/total marks) * 100 => percentage
# total marks obtained / 3  => Average

a = int(input('physics marks '))
b = int(input('chemistry marks '))
c = int(input('biology marks '))

perc = (a + b + c) / 3
print('percentage: ', perc)

if perc >70 and perc <= 80:
    print('Grade C')
elif perc > 80 and perc <= 90:
    print('Grade B')
elif perc > 90 and perc <=100:
    print('Grade A')
    
else:
    print('Not Qualified')
```
